```python
section_data = data["data"][0]["sectionData"]
```

```python


def extract_all_texts(response: dict) -> str:
    texts = []

    def collect(obj):
        if isinstance(obj, dict):
            if "text" in obj and isinstance(obj["text"], str):
                texts.append(obj["text"].strip())
            for v in obj.values():
                collect(v)
        elif isinstance(obj, list):
            for item in obj:
                collect(item)

    collect(response)
    return " | ".join(t for t in texts if t)
extract_all_texts(section_data)
```

```python

def _has_overseas_tracking(data: dict) -> bool:
    """Đệ quy tìm utility_title=shipping_overseas_info trong foldTracking."""
    if isinstance(data, dict):
        for k, v in data.items():
            if k == "origin":
                try:
                    origin = json.loads(v)
                    if origin.get("utility_title") == "shipping_overseas_info":
                        return True
                except (json.JSONDecodeError, TypeError):
                    pass
            if _has_overseas_tracking(v):
                return True
    elif isinstance(data, list):
        for item in data:
            if _has_overseas_tracking(item):
                return True
    return False
```

```python
import re

def parse_shipping_info(response: dict) -> dict:
    result = {
        "delivery_charge": 0,
        "delivery_company": None,
        "shipping_overseas": False,
        "delivery_text": None,
    }

    data = response.get("data", [])
    if not data:
        return result

    section_data = data[0].get("sectionData", {})
    result["delivery_text"] = extract_all_texts(section_data)
    for sub in section_data.get("detailSubSections", []):
        sub_data = sub.get("data") or {}

        # delivery_company từ secondaryDisplays
        for d in sub_data.get("secondaryDisplays") or []:
            text = d.get("text", "").strip()
            if text:
                result["delivery_company"] = text

        # delivery_charge từ basicDisplays
        for d in sub_data.get("basicDisplays") or []:
            text = d.get("text", "").strip()
            if "배송비" in text:
                # extract số: "배송비 3,000원" -> "3000"
                match = re.search(r"([\d,]+)원", text)
                if match:
                    result["delivery_charge"] = int(match.group(1).replace(",", ""))

            # shipping_overseas
            result["shipping_overseas"] = "shipping_overseas_info" in str(section_data)



    return result
parse_shipping_info(data)
```
