```python
import os
import shutil
try:
    import winreg
except ImportError:
    winreg = None
from pathlib import Path


def get_browser_executable(browser_name: str = "chrome") -> str:
    browser_name = browser_name.lower()

    registry_keys = {
        "chrome": [
            r"SOFTWARE\Microsoft\Windows\CurrentVersion\App Paths\chrome.exe",
            r"SOFTWARE\WOW6432Node\Microsoft\Windows\CurrentVersion\App Paths\chrome.exe",
        ],
        "edge": [
            r"SOFTWARE\Microsoft\Windows\CurrentVersion\App Paths\msedge.exe",
            r"SOFTWARE\WOW6432Node\Microsoft\Windows\CurrentVersion\App Paths\msedge.exe",
        ],
        "firefox": [
            r"SOFTWARE\Microsoft\Windows\CurrentVersion\App Paths\firefox.exe",
            r"SOFTWARE\WOW6432Node\Microsoft\Windows\CurrentVersion\App Paths\firefox.exe",
        ],
    }

    exe_names = {
        "chrome": "chrome.exe",
        "edge": "msedge.exe",
        "firefox": "firefox.exe",
    }

    if browser_name not in registry_keys:
        raise ValueError("browser_name must be: chrome, edge, or firefox")

    # 1. Tìm bằng Registry
    for key_path in registry_keys[browser_name]:
        for root in (winreg.HKEY_CURRENT_USER, winreg.HKEY_LOCAL_MACHINE):
            try:
                with winreg.OpenKey(root, key_path) as key:
                    path, _ = winreg.QueryValueEx(key, "")
                    if path and Path(path).exists():
                        return path
            except FileNotFoundError:
                pass

    # 2. Fallback: tìm trong PATH
    path = shutil.which(exe_names[browser_name])
    if path:
        return path

    raise FileNotFoundError(f"Could not find {browser_name} executable.")


print(get_browser_executable("chrome"))
print(get_browser_executable("edge"))
print(get_browser_executable("firefox"))
```

```python
async def solve_and_confirm(page, solver, max_retries=1):
    for attempt in range(max_retries):
        try:
            await solver.solveCaptcha()
            
            # Solved → click 확인 (OK)
            await page.click("button.c-button--primary")
            print("[INFO] Clicked 확인")
            return True
            
        except Exception as e:
            print(f"[WARN] Attempt {attempt+1} failed: {e}")
            
            if attempt < max_retries - 1:
                # Refresh và thử lại
                await page.reload()
                await page.wait_for_timeout(2000)
                print(f"[INFO] Retrying... ({attempt+2}/{max_retries})")
    
    raise Exception("Failed after max retries")


# Dùng
await solve_and_confirm(page, self)
```
